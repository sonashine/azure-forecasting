```python
import numpy as np
import pandas as pd
from neuralforecast import NeuralForecast
from neuralforecast.models import LSTM
import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv('../challenge_data/vm_cpu_readings.csv.gz')
```

```python
g = df.groupby("timestamp")
active = g["vm_id"].nunique()
agg_cpu = g["avg_cpu"].mean()
```

```python
train_cpu, hold_cpu = agg_cpu.iloc[:110], agg_cpu.iloc[110:]

df = pd.DataFrame({"unique_id": "agg_cpu",
                      "ds": np.arange(len(train_cpu)),
                      "y": train_cpu.values})
```

```python
def make_nf(layers=1, hidden=16, seed=1, dropout=0.0):
    return NeuralForecast(
        models=[LSTM(h=24, input_size=24, encoder_hidden_size=hidden, encoder_n_layers=layers,
                     max_steps=300, scaler_type="standard", random_seed=seed)],
        freq=1)
```

```python
cv_lstm = make_nf(layers=3).cross_validation(df=df, n_windows=10, step_size=4, refit=True)
print(sorted(cv_lstm["cutoff"].unique()))   
print(cv_lstm["ds"].max())      
```

```python
e = cv_lstm["y"] - cv_lstm["LSTM"]
print("MAE", e.abs().mean(), "RMSE", (e**2).mean() ** 0.5)
```
